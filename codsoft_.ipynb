```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
num_cols = df.select_dtypes(include='number').columns

df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df[num_cols]
```

```python
cat_cols = df.select_dtypes(include='object').columns

df[cat_cols] = df[cat_cols].fillna("Unknown")
df[cat_cols]
```

```python
df.isnull().sum()
```

```python
df.to_csv("Amazon_Sales_Cleaned.csv", index=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#1. Total Sales
df['Amount'].sum()
```

```python
#2. Average Order Value
df['Amount'].mean()
```

```python
#3. Total Quantity Sold
df['Qty'].sum()
```

```python
#4. Order Status Distribution
df['Status'].value_counts()
```

```python
#4. Order Status Distribution(Visualization)
plt.figure(figsize=(10,5))
sns.countplot(data=df, y="Status", order=df["Status"].value_counts().index)
plt.title("Order Status Distribution")
plt.show()
```

```python
#5. Top 10 Product Categories
df['Category'].value_counts().head(10)
```

```python
#5. Top 10 Product Categories(Visualization)
plt.figure(figsize=(8,5))
sns.countplot(data=df,
              y="Category",
              order=df["Category"].value_counts().index)
plt.title("Top Categories")
plt.show()
```

```python
#6. Which Size is Sold Most?
df["Size"].value_counts()
```

```python
#6. Which Size is Sold Most?(Visualization)
plt.figure(figsize=(8,5))
sns.countplot(data=df,y='Size',order=df["Size"].value_counts().index)
plt.title("Size sold")
plt.show()
```

```python
#7. Top 10 States by Sales
state_sales = df.groupby("ship-state")["Amount"].sum().sort_values(ascending=False)

state_sales.head(10)
```

```python
#8. Amazon vs Merchant Fulfilment
df["Fulfilment"].value_counts()
```

```python
##8. Amazon vs Merchant Fulfilment(Visualization)
plt.figure(figsize=(6,6))
df["Fulfilment"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.ylabel("")
plt.show()
```

```python
df["Courier Status"].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.select_dtypes(include="number").corr(),
            annot=True,
            cmap="Blues")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df["Amount"])
plt.show()
```

```python
#Sales by Category
category_sales = df.groupby("Category")["Amount"].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
category_sales.plot(kind="bar", color="skyblue")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
#2.Monthly Sales Trend
df["Date"] = pd.to_datetime(df["Date"])

monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Amount"].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot(marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
#3.Fulfillment Method Distribution
fulfillment = df["Fulfilment"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    fulfillment,
    labels=fulfillment.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Fulfillment Method Distribution")
plt.show()
```
